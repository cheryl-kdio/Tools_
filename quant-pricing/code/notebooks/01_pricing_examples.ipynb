# Exemples de pricing
Données fictives ; taux en décimal, temps en années.

```python
from quant_pricing.rates import estr_interest, euribor_interest
from quant_pricing.curves import DiscountCurve
from quant_pricing.bonds import dirty_price, yield_to_maturity
from quant_pricing.irs import irs_npv, par_swap_rate
from quant_pricing.trs import trs_npv, total_return
from quant_pricing.options import black_scholes, greeks
import math
c = DiscountCurve([0,1,2], [1, math.exp(-.03), math.exp(-.06)])
```

## Capitalisation €STR et coupon Euribor

```python
estr_interest(1e6, [.03,.031], [3,1]), euribor_interest(1e6,.03,.25)
```

## Obligation fixe

```python
flows = [(1,5),(2,105)]
dirty_price(flows,c), yield_to_maturity(flows,100,1)
```

## IRS — courbe unique pour cet exemple uniquement

```python
fixed = [(1,1),(2,1)]
floating = [(1,1,c.forward(0,1,1)),(2,1,c.forward(1,2,1))]
par_swap_rate(fixed,floating,c), irs_npv(1e6,.025,fixed,floating,c)
```

## TRS — scénario de rendement fourni, pas une prévision de marché

```python
periods = [(1,1,total_return(100,105,2),.03)]
trs_npv(1e6,periods,c,.005)
```

## Options européennes

```python
black_scholes(100,100,1,.05,.2), black_scholes(100,100,1,.05,.2,"put"), greeks(100,100,1,.05,.2)
```
